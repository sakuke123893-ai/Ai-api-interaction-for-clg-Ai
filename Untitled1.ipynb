```python
def ask_llm(user_prompt, system_prompt=None, temperature=0.7, model=MODEL):
    """Send a single-turn chat request to Groq and return the reply text."""
    messages = []
    if system_prompt:
        messages.append({"role": "system", "content": system_prompt})
    messages.append({"role": "user", "content": user_prompt})

    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
    )
    return response


'''print("Ai college idea")
n = int(input("Enter maximum number of questions: "))

for i in range(n):
    print("\nQuestion", i + 1)
    question = input("Enter your college-related question: ")

    user_prompt = f"""
Give us 3 suggestions about this college-related question:

{question}
"""
    system_prompt = "You are a helpful college academic and career advisor."
    temperature = 0.7

    result = ask_llm(
        user_prompt=user_prompt,
        system_prompt=system_prompt,
        temperature=temperature,
    )

    print("\nAI Suggestions:")
    print(result.choices[0].message.content)
    print("\n")'''
question = choice
print("AI College Idea Generator")

question = input("\nEnter your college-related question: ")

while True:

    user_prompt = f"""
Give min 3 creative suggestions about this college-related question or mentioned to give :

{question}
"""

    result = ask_llm(
        user_prompt,
        system_prompt,
        temperature
    )

    print("\nAI Suggestions:")
    print(result.choices[0].message.content)

    question = input("\nDo you have another question? "
                     "Type your question or 'no' to stop: ")

    if question.lower() == "no":
        print("\nProgram stopped.")
        break
```
